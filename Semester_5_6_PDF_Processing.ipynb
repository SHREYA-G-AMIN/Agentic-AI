**Read the PDF**

Install/import the PDF reader:

```python
!pip install pypdf
from pypdf import PdfReader

reader = PdfReader(pdf_path)

print("Number of pages:", len(reader.pages))
```

Convert PDF into text and display it

```python
pdf_text = ""

for page in reader.pages:
    pdf_text += page.extract_text() + "\n"

print(pdf_text)
```

Clean the text

```python
import re

clean_text = re.sub(r'\s+', ' ', pdf_text).strip()

print(clean_text)
```

Count characters

```python
print("Total characters:", len(clean_text))
```

Split into 200-character chunks

```python
chunk_size = 200

chunks = [
    clean_text[i:i + chunk_size]
    for i in range(0, len(clean_text), chunk_size)
]

print("Total chunks:", len(chunks))

for n, chunk in enumerate(chunks[:20], start=1):
    print(f"{n:>4}: {chunk}")
```

Save all chunks

```python
with open("chunks.txt", "w", encoding="utf-8") as f:
    f.write("\n".join(chunks))
```
